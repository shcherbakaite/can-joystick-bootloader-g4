# CAN Bootloader

## Bootloader operation

1. Controller starts excuting the bootloader code on power-up 
2. Bootloader configures CAN periphiral to listen for message to enter update mode
5. Each received message is stored in a buffer
6. The buffer has a limited size and must be flashed
7. During flashing controller will be unresponsive, so the send rate must not be faster than the slowest stall time of the flash buffer write
